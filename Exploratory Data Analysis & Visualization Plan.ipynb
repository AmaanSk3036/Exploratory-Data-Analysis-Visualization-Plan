# HISTOGRAM UNDERSTAND DISTRIBUTION

# BOX PLOT : DETECT POTENTIAL OUTLIERS

```python
# Examine unit prices
prices = df["UnitPrice"].dropna()

plt.figure(figsize=(10, 4))
sns.boxplot(x=prices)

plt.title("Box Plot of Unit Prices")
plt.xlabel("Unit Price (£)")
plt.tight_layout()
plt.show()
```

# BAR CHART : COMPARE COUNTRIES 

```python
# Create a copy for exploratory analysis
eda = df.copy()

# Convert invoice dates to datetime
eda["InvoiceDate"] = pd.to_datetime(
    eda["InvoiceDate"],
    errors="coerce"
)

# Calculate line-item value
eda["LineValue"] = eda["Quantity"] * eda["UnitPrice"]

# Display the calculated values
display(eda[["Quantity", "UnitPrice", "LineValue"]].head())
```

```python
# Sum line-item values by country
country_sales = (
    eda.dropna(subset=["Country", "LineValue"])
       .groupby("Country")["LineValue"]
       .sum()
       .sort_values(ascending=False)
       .head(10)
       .sort_values()
)

# Draw a horizontal bar chart
plt.figure(figsize=(10, 6))
country_sales.plot(kind="barh")

plt.title("Top 10 Countries by Aggregate Line-Item Value")
plt.xlabel("Aggregate Line-Item Value (£)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

# LINE CHART : EXAMINE MONTHLY TRENDS 

```python
# Extract the month from each invoice date
eda["Month"] = eda["InvoiceDate"].dt.to_period("M")

# Calculate aggregate line-item value by month
monthly_value = (
    eda.dropna(subset=["Month", "LineValue"])
       .groupby("Month")["LineValue"]
       .sum()
)

# Plot the monthly trend
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_value.index.astype(str),
    monthly_value.values,
    marker="o"
)

plt.title("Monthly Aggregate Line-Item Value")
plt.xlabel("Month")
plt.ylabel("Aggregate Line-Item Value (£)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# SCATTER PLOT : EXAMINE RELATIONSHIP  

```python
# Use rows with valid numeric values
scatter_data = eda[
    ["Quantity", "UnitPrice"]
].dropna()

# Sample rows to keep the chart readable
scatter_sample = scatter_data.sample(
    n=min(5000, len(scatter_data)),
    random_state=42
)

# Draw the scatter plot
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=scatter_sample,
    x="Quantity",
    y="UnitPrice",
    alpha=0.4
)

plt.title("Quantity vs Unit Price")
plt.xlabel("Quantity")
plt.ylabel("Unit Price (£)")
plt.tight_layout()
plt.show()
```

# CORRELATION HEARMAP : EXAMINE NUMERIC RELATIONSHIP 

```python
# Select relevant numeric variables
numeric_data = eda[
    ["Quantity", "UnitPrice", "LineValue"]
].dropna()

# Calculate Pearson correlation coefficients
correlation = numeric_data.corr()

# Plot the heatmap
plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Numeric Variables")
plt.tight_layout()
plt.show()
```

# DESCRIPTIVE STATISTICS 

```python
# Select the variables for descriptive analysis
variables = ["Quantity", "UnitPrice", "LineValue"]

# Calculate central tendency and spread
summary = eda[variables].describe(
    percentiles=[0.25, 0.50, 0.75, 0.95]
).T

# Add median, mode, range, variance, and IQR
for column in variables:
    values = eda[column].dropna()

    summary.loc[column, "median"] = values.median()
    summary.loc[column, "mode"] = (
        values.mode().iloc[0] if not values.mode().empty else np.nan
    )
    summary.loc[column, "range"] = values.max() - values.min()
    summary.loc[column, "variance"] = values.var()
    summary.loc[column, "IQR"] = (
        values.quantile(0.75) - values.quantile(0.25)
    )

display(summary.round(2))
```

# MONTHLY REVENUE & TRANSACTION VOLUME 

```python

# Prepare a subset for positive-value sales analysis
sales_eda = eda[
    (eda["Quantity"] > 0) &
    (eda["UnitPrice"] > 0) &
    (~eda["InvoiceNo"].astype(str).str.startswith("C"))
].copy()

# Create monthly summary
sales_eda["Month"] = sales_eda["InvoiceDate"].dt.to_period("M")

monthly_summary = sales_eda.groupby("Month").agg(
    Revenue=("LineValue", "sum"),
    Invoices=("InvoiceNo", "nunique")
).reset_index()

monthly_summary["Month"] = monthly_summary["Month"].astype(str)

# Draw the chart
fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.bar(
    monthly_summary["Month"],
    monthly_summary["Revenue"]
)
ax1.set_xlabel("Month")
ax1.set_ylabel("Positive-Value Sales (£)")
ax1.tick_params(axis="x", rotation=45)

ax2 = ax1.twinx()
ax2.plot(
    monthly_summary["Month"],
    monthly_summary["Invoices"],
    marker="o"
)
ax2.set_ylabel("Number of Unique Invoices")

plt.title("Monthly Sales Value vs Invoice Volume")
plt.tight_layout()
plt.show()
```

# DAY OF WEEK SALES COMPARISON 

```python
# Extract weekday name and weekday number
sales_eda["Weekday"] = sales_eda["InvoiceDate"].dt.day_name()
sales_eda["WeekdayNumber"] = sales_eda["InvoiceDate"].dt.dayofweek

# Aggregate by weekday
weekday_sales = (
    sales_eda.groupby(["WeekdayNumber", "Weekday"])["LineValue"]
    .sum()
    .reset_index()
    .sort_values("WeekdayNumber")
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=weekday_sales,
    x="Weekday",
    y="LineValue"
)

plt.title("Sales Value by Day of the Week")
plt.xlabel("Weekday")
plt.ylabel("Positive-Value Sales (£)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

# REVENUE CONCENTARTION USING PARETO CHART 

```python
# Aggregate sales value by product description
product_sales = (
    sales_eda.dropna(subset=["Description"])
    .groupby("Description")["LineValue"]
    .sum()
    .sort_values(ascending=False)
)

# Select the top 15 products
top_products = product_sales.head(15)

# Calculate cumulative percentage across ALL products
cumulative_percent = (
    product_sales.cumsum() / product_sales.sum() * 100
)

# Keep the cumulative percentages for the top 15
top_cumulative = cumulative_percent.loc[top_products.index]

fig, ax1 = plt.subplots(figsize=(13, 6))

ax1.bar(
    range(len(top_products)),
    top_products.values
)
ax1.set_xticks(range(len(top_products)))
ax1.set_xticklabels(top_products.index, rotation=75, ha="right")
ax1.set_ylabel("Sales Value (£)")
ax1.set_xlabel("Product Description")

ax2 = ax1.twinx()
ax2.plot(
    range(len(top_products)),
    top_cumulative.values,
    marker="o"
)
ax2.set_ylabel("Cumulative Share of All Product Sales (%)")
ax2.set_ylim(0, 105)

plt.title("Pareto Analysis: Top 15 Products by Sales Value")
plt.tight_layout()
plt.show()
```

# CUSTOMER PURCHASE FREQUENCY 

```python
# Count unique invoices for each identified customer
customer_frequency = (
    sales_eda.dropna(subset=["CustomerID"])
    .groupby("CustomerID")["InvoiceNo"]
    .nunique()
)

plt.figure(figsize=(10, 5))
sns.histplot(
    customer_frequency,
    bins=30,
    kde=True
)

plt.title("Distribution of Customer Purchase Frequency")
plt.xlabel("Number of Unique Invoices per Customer")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

# AVERAGE INVOICE VALUE DISTRIBUTION 

```python
# Calculate total positive-value sales per invoice
invoice_values = (
    sales_eda.groupby("InvoiceNo")["LineValue"]
    .sum()
)

plt.figure(figsize=(10, 5))
sns.histplot(
    invoice_values,
    bins=50,
    kde=True
)

plt.axvline(
    invoice_values.median(),
    linestyle="--",
    label=f"Median: £{invoice_values.median():.2f}"
)

plt.title("Distribution of Invoice Values")
plt.xlabel("Invoice Value (£)")
plt.ylabel("Number of Invoices")
plt.legend()
plt.tight_layout()
plt.show()
```

# COUNTRY COMPARISON : SALES VALUE VS NUMBER OF INVOICES 

```python
# Summarize sales by country
country_summary = sales_eda.groupby("Country").agg(
    Revenue=("LineValue", "sum"),
    Invoices=("InvoiceNo", "nunique"),
    Customers=("CustomerID", "nunique")
).reset_index()

# Select the 15 countries with the highest sales value
country_sample = country_summary.nlargest(15, "Revenue")

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=country_sample,
    x="Invoices",
    y="Revenue",
    size="Customers",
    sizes=(50, 500),
    hue="Revenue",
    legend=False
)

# Label each point with its country
for _, row in country_sample.iterrows():
    plt.annotate(
        row["Country"],
        (row["Invoices"], row["Revenue"]),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=8
    )

plt.title("Country-Level Sales Value vs Invoice Count")
plt.xlabel("Number of Unique Invoices")
plt.ylabel("Positive-Value Sales (£)")
plt.tight_layout()
plt.show()
```

# HOURLY SALES PATTERN 

```python
# Extract the hour from invoice timestamps
sales_eda["Hour"] = sales_eda["InvoiceDate"].dt.hour

# Calculate total sales value by hour
hourly_sales = (
    sales_eda.groupby("Hour")["LineValue"]
    .sum()
    .reindex(range(24), fill_value=0)
)

plt.figure(figsize=(11, 5))

plt.plot(
    hourly_sales.index,
    hourly_sales.values,
    marker="o"
)

plt.title("Sales Value by Hour of Day")
plt.xlabel("Hour of Day (24-Hour Clock)")
plt.ylabel("Positive-Value Sales (£)")
plt.xticks(range(24))
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

# WEEKDAY & HOUR HEATMAP 

```python
# Create weekday and hour fields
sales_eda["Weekday"] = sales_eda["InvoiceDate"].dt.day_name()
sales_eda["WeekdayNumber"] = sales_eda["InvoiceDate"].dt.dayofweek

# Aggregate sales by weekday and hour
time_heatmap = sales_eda.pivot_table(
    index="WeekdayNumber",
    columns="Hour",
    values="LineValue",
    aggfunc="sum",
    fill_value=0
)

# Ensure all 24 hours are represented
time_heatmap = time_heatmap.reindex(
    columns=range(24),
    fill_value=0
)

plt.figure(figsize=(14, 6))

sns.heatmap(
    time_heatmap,
    cmap="YlOrRd"
)

plt.title("Sales Value by Weekday and Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Weekday (0 = Monday, 6 = Sunday)")
plt.tight_layout()
plt.show()
```

# PRODUCT QUANTITY DISTRIBUTION BY WEEKDAY 

```python
# Use a manageable sample for visualization
quantity_sample = sales_eda[
    sales_eda["Quantity"] <= sales_eda["Quantity"].quantile(0.99)
].copy()

quantity_sample = quantity_sample.sample(
    n=min(10000, len(quantity_sample)),
    random_state=42
)

# Plot quantity distributions by weekday
plt.figure(figsize=(11, 6))

sns.violinplot(
    data=quantity_sample,
    x="WeekdayNumber",
    y="Quantity",
    inner="quartile",
    cut=0
)

plt.title("Distribution of Item Quantities by Weekday")
plt.xlabel("Weekday (0 = Monday, 6 = Sunday)")
plt.ylabel("Quantity per Line Item")
plt.tight_layout()
plt.show()
```

# RETURN & CANCELLATION OVERVIEW 

```python
# Identify invoices whose identifiers start with C
is_cancellation = eda["InvoiceNo"].astype(str).str.startswith("C")

# Count cancellation-identified and other invoice records
invoice_types = pd.Series({
    "Invoice IDs starting with C": is_cancellation.sum(),
    "Other line-item records": (~is_cancellation).sum()
})

# Count positive, negative, and zero quantities
quantity_types = pd.Series({
    "Positive quantity": (eda["Quantity"] > 0).sum(),
    "Negative quantity": (eda["Quantity"] < 0).sum(),
    "Zero quantity": (eda["Quantity"] == 0).sum()
})

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

invoice_types.plot(kind="bar", ax=axes[0])
axes[0].set_title("Records by Invoice Identifier")
axes[0].set_ylabel("Number of Rows")
axes[0].tick_params(axis="x", rotation=20)

quantity_types.plot(kind="bar", ax=axes[1])
axes[1].set_title("Records by Quantity Sign")
axes[1].set_ylabel("Number of Rows")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```
